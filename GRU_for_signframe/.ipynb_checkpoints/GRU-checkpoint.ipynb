```python


import os
import numpy as np
import tensorflow as tf

from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
```

```python
#cnn feature extraction
cnn=MobileNetV2(
    weights="imagenet",
    include_top=False,
    pooling="avg"
)

cnn.trainable=False
```

```python
#iterating through the dataset

import os

#dataset path
DATASET_ROOT=r"C:\Users\darsh\Documents\ML_Datasets\wlasl\preprocessing"

dataset=[]

for split in ["train", "val", "test"]:

    print(f"\n========== {split.upper()} ==========")

    frames_path=os.path.join(
        DATASET_ROOT,
        split,
        "frames"
    )

    for word in os.listdir(frames_path):
        word_path=os.path.join(
            frames_path,
            word
        )

        if not os.path.isdir(word_path):
            continue

        for sequence in os.listdir(word_path):
            sequence_path=os.path.join(
                word_path,
                sequence
            )

            if not os.path.isdir(sequence_path):
                continue

            #all image paths
            image_paths=[]

            for image in sorted(os.listdir(sequence_path)):
                if image.lower().endswith(".jpg"):
                    image_path=os.path.join(
                        sequence_path,
                        image
                    )

                    image_paths.append(image_path)

            dataset.append({
                "split":split,
                "word":word,
                "sequence":sequence,
                "image_paths":image_paths
            })

print("Total sequence: ",len(dataset))

print(dataset[0])
```

```python
import tensorflow as tf
import numpy as np
def load_sequence_images(image_paths):
    frames=[]

    for image_path in image_paths:
        image=tf.keras.utils.load_img(
            image_path,
            target_size=(224,224)
        )

        image=tf.keras.utils.img_to_array(image)

        frames.append(image)

    return np.array(frames, dtype=np.float32)

```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

FEATURES_ROOT=r"E:\signframe\GRU_for_signframe\cnn_extracted_features"

def extract_cnn_features(item):

    #load all images in this sequence
    frames=load_sequence_images(
        item["image_paths"]
    )

    #preprocess for MobileNetV2
    frames=preprocess_input(frames)

    #extract CNN features
    features=cnn.predict(
        frames,
        verbose=0
    )

    #create directory
    save_dir=os.path.join(
        FEATURES_ROOT,
        item["split"],
        item["word"]
    )

    os.makedirs(save_dir, exist_ok=True)

    #create path
    save_path=os.path.join(
        save_dir,
        f"{item['sequence']}.npy"
    )

    #save extracted features
    np.save(save_path, features) 

    return features
```

```python
def process_split(split_name):
    X=[]
    y=[]

    for item in dataset:

        if item["split"]!=split_name:
            continue

        features=extract_cnn_features(
            item
        )

        #store features
        X.append(features)

        #store word label
        y.append(item["word"])

    return X,y
```

```python
#end of CNN, features are stored in .npy file on the disk in cnn_extracted_features folders

from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

X_train, y_train = process_split("train")

X_val, y_val = process_split("val")

X_test, y_test = process_split("test")

print("Training sequences:", len(X_train))
print("Validation sequences:", len(X_val))
print("Testing sequences:", len(X_test))
```

```python
#Loading .npy files from disk

import os
import numpy as np

FEATURES_ROOT = r"E:\signframe\GRU_for_signframe\cnn_extracted_features"


def load_feature_split(split_name):

    X = []
    y = []

    split_path = os.path.join(
        FEATURES_ROOT,
        split_name
    )

    # Each folder represents one word/class
    for word in sorted(os.listdir(split_path)):

        word_path = os.path.join(
            split_path,
            word
        )

        if not os.path.isdir(word_path):
            continue

        # Each .npy file represents one sign sequence
        for filename in sorted(os.listdir(word_path)):

            if not filename.endswith(".npy"):
                continue

            feature_path = os.path.join(
                word_path,
                filename
            )

            # Load already-extracted CNN features
            features = np.load(feature_path)

            X.append(features)

            # Folder name is the word label
            y.append(word)

    return X, y
```

```python
X_train, y_train = load_feature_split("train")

X_val, y_val = load_feature_split("val")

X_test, y_test = load_feature_split("test")
```

```python
print("Training sequences:", len(X_train))
print("Validation sequences:", len(X_val))
print("Testing sequences:", len(X_test))
```

```python
# Get the 100 word classes

class_names = sorted(set(y_train))

print("Number of classes:", len(class_names))
print(class_names)
```

```python
# Create label -> integer mapping

label_to_id = {
    label: index
    for index, label in enumerate(class_names)
}

# Create integer -> label mapping
# Useful later when interpreting predictions

id_to_label = {
    index: label
    for label, index in label_to_id.items()
}

print("All 100 label mappings:")

for index in range(100):
    print(index, "->", id_to_label[index])
```

```python
# Convert string labels to integer class IDs

y_train_encoded = np.array([
    label_to_id[label]
    for label in y_train
])

y_val_encoded = np.array([
    label_to_id[label]
    for label in y_val
])

y_test_encoded = np.array([
    label_to_id[label]
    for label in y_test
])
```

```python
print("Training labels shape:", y_train_encoded.shape)
print("Validation labels shape:", y_val_encoded.shape)
print("Testing labels shape:", y_test_encoded.shape)

print("\nFirst 10 training labels:")
print(y_train_encoded[:1440])
```

```python
# Verify encoding

for i in range(1440):
    original_label = y_train[i]
    encoded_label = y_train_encoded[i]
    decoded_label = id_to_label[encoded_label]

    print(
        original_label,
        "->",
        encoded_label,
        "->",
        decoded_label
    )
```

```python
# Convert lists of sequences into 3D NumPy arrays so that GRU can accept it

X_train = np.stack(X_train)
X_val = np.stack(X_val)
X_test = np.stack(X_test)

print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)
```

```python
print("X_train dtype:", X_train.dtype)
print("X_val dtype:", X_val.dtype)
print("X_test dtype:", X_test.dtype)
```

```python
# Build the GRU model

import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Input(
        shape=(16, 1280)
    ),

    tf.keras.layers.GRU(
        128
    ),

    tf.keras.layers.Dense(
        100,
        activation="softmax"
    )
])

model.summary()
```

```python
# Compile the GRU model

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

```python
#creating a checkpoint for saving the best model

checkpoint = tf.keras.callbacks.ModelCheckpoint(
    "signframe_gru_best.keras",
    monitor="val_accuracy",
    save_best_only=True,
    mode="max",
    verbose=1
)
```

```python
history = model.fit(
    X_train,
    y_train_encoded,

    validation_data=(
        X_val,
        y_val_encoded
    ),

    epochs=20,
    batch_size=32,

    callbacks=[checkpoint]
)
```

```python
best_model = tf.keras.models.load_model(
    "signframe_gru_best.keras"
)

for i in range(20):
    test_loss, test_accuracy = best_model.evaluate(
    X_test,
    y_test_encoded,
    batch_size=32
    )
    print("Epoch:", i)
    print("Test loss:", test_loss)
    print("Test accuracy:", test_accuracy)
```
